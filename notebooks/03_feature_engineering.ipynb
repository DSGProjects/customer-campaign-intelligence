# Customer & Campaign Intelligence — Feature Engineering

**Proyecto:** Segmentación de clientes y análisis de respuesta a campañas de marketing.

**Objetivo de este notebook:** crear variables derivadas a partir del dataset limpio
(`02_data_cleaning.ipynb`), orientadas a las preguntas de negocio del proyecto:
valor del cliente, comportamiento de compra y respuesta a campañas.

**Entrada:** `data/processed/marketing_campaign_clean.csv`
**Salida:** `data/processed/marketing_campaign_features.csv`

```python
df["Dependientes"] = df["Kidhome"] + df["Teenhome"]
```

```python
fecha_referencia = df["Dt_Customer"].max()
print(fecha_referencia)
```

```python
df["Antiguedad_Cliente"] = fecha_referencia - df["Dt_Customer"]
```

```python
df["Antiguedad_Cliente"] = df["Antiguedad_Cliente"].dt.days
```

```python
df["Segmento_Valor"] = pd.cut(
    df["Gasto_Total"],
    bins=[-1, 110.65, 813.20, df["Gasto_Total"].max()],
    labels=["Bajo", "Medio", "Alto"]
)
```

## Feature Engineering — Resumen

A partir del dataset limpio (2,236 filas × 27 columnas), se crearon las 
siguientes variables:

1. **Edad:** 2014 (año más reciente en `Dt_Customer`) − `Year_Birth`.
2. **Gasto_Total:** suma del gasto en las 6 categorías de producto 
   (`MntWines`, `MntFruits`, `MntMeatProducts`, `MntFishProducts`, 
   `MntSweetProducts`, `MntGoldProds`).
3. **Compras_Totales:** suma de compras por los 3 canales 
   (`NumWebPurchases`, `NumCatalogPurchases`, `NumStorePurchases`).
4. **Campañas_Aceptadas:** suma de las 5 campañas anteriores aceptadas 
   (`AcceptedCmp1` a `AcceptedCmp5`).
5. **Dependientes:** suma de `Kidhome` + `Teenhome`.
6. **Antiguedad_Cliente:** días transcurridos entre `Dt_Customer` y la 
   fecha más reciente del dataset (2014-06-29).
7. **Segmento_Valor:** clasificación en Bajo/Medio/Alto según percentiles 
   (33% y 66%) de `Gasto_Total`.

**Resultado final:** 2,236 filas × 34 columnas, guardado en 
`data/processed/marketing_campaign_features.csv`.
